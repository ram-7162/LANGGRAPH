```python
from langgraph.graph import StateGraph, START, END
from typing import TypedDict
```

```python
class BatsmanState(TypedDict):
    runs: int
    balls: int
    fours: int 
    sixes: int 

    sr: float
    bpb: float
    boundary_percent: float
    summary : str
```

```python
def calculate_sr(state : BatsmanState) -> BatsmanState:
    runs = state['runs']
    balls = state['balls']
    sr = runs/balls.float()
    state['sr'] = sr
    return state
```

```python
def calculate_bpb(state : BatsmanState) -> BatsmanState:
    fours = state['fours']
    sixes = state["sixes"]
    total_boundary = (fours + sixes )
    balls = state['balls']
    bpb =  total_boundary/balls.float()
    state['bpb'] = bpb
    return state

```

```python
def calculate_boundary_percent(state : BatsmanState) -> BatsmanState:
    fours = state['fours']
    sixes = state["sixes"]
    total_boundary = (fours*4 + sixes*6 )
    runs = state['runs']
    bp = (total_boundary * 100)/runs.float()
    state['boundary_percent'] = bp
    return state
```

```python
def summary(state : BatsmanState) -> BatsmanState:
    summary = f"""
    strike_rate = {state['sr']}
    boundary_per_ball = {state['bpb']}
    boundary_percent = {state['boundary_percent']}"""
    state['summary'] = summary
    return state
```

```python
graph = StateGraph(BatsmanState)

graph.add_node('sr_calculator', calculate_sr)
graph.add_node('bpb_calculator', calculate_bpb)
graph.add_node('boundary_percent_calculator', calculate_boundary_percent)
graph.add_node('summary', summary)


graph.add_edge(START, "sr_calculator")
graph.add_edge(START, "bpb_calculator")
graph.add_edge(START, "boundary_percent_calculator")
graph.add_edge("sr_calculator", "summary")
graph.add_edge("bpb_calculator", "summary")
graph.add_edge("boundary_percent_calculator", "summary")
graph.add_edge("summary", END)
workflow = graph.compile()
```
